```python
class No:
    def __init__(self, valor):
        self.valor = valor
        self.esquerda = None
        self.direita = None
        self.altura = 1


class ArvoreAVL:
    def __init__(self):
        self.raiz = None

    # Retorna a altura de um nó
    def altura(self, no):
        if no is None:
            return 0
        return no.altura

    # Calcula o fator de balanceamento
    def fator_balanceamento(self, no):
        if no is None:
            return 0

        return self.altura(no.esquerda) - self.altura(no.direita)

    # Atualiza a altura do nó
    def atualizar_altura(self, no):
        no.altura = 1 + max(
            self.altura(no.esquerda),
            self.altura(no.direita)
        )

    # Rotação para a direita
    def rotacao_direita(self, no):
        filho = no.esquerda
        temp = filho.direita

        filho.direita = no
        no.esquerda = temp

        self.atualizar_altura(no)
        self.atualizar_altura(filho)

        return filho

    # Rotação para a esquerda
    def rotacao_esquerda(self, no):
        filho = no.direita
        temp = filho.esquerda

        filho.esquerda = no
        no.direita = temp

        self.atualizar_altura(no)
        self.atualizar_altura(filho)

        return filho

    # Inserir um valor
    def inserir_no(self, no, valor):

        # Se chegou em uma posição vazia,
        # cria um novo nó
        if no is None:
            return No(valor)

        # Inserção igual à BST
        if valor < no.valor:
            no.esquerda = self.inserir_no(
                no.esquerda, valor
            )

        elif valor > no.valor:
            no.direita = self.inserir_no(
                no.direita, valor
            )

        else:
            print("Valor já existe!")
            return no

        # Atualiza a altura
        self.atualizar_altura(no)

        # Calcula o fator de balanceamento
        fb = self.fator_balanceamento(no)

        # ==========================================
        # CASO LL
        # ==========================================

        if fb > 1 and valor < no.esquerda.valor:
            return self.rotacao_direita(no)

        # ==========================================
        # CASO RR
        # ==========================================

        if fb < -1 and valor > no.direita.valor:
            return self.rotacao_esquerda(no)

        # ==========================================
        # CASO LR
        # ==========================================

        if fb > 1 and valor > no.esquerda.valor:
            no.esquerda = self.rotacao_esquerda(
                no.esquerda
            )

            return self.rotacao_direita(no)

        # ==========================================
        # CASO RL
        # ==========================================

        if fb < -1 and valor < no.direita.valor:
            no.direita = self.rotacao_direita(
                no.direita
            )

            return self.rotacao_esquerda(no)

        return no

    # Método público para inserir
    def inserir(self, valor):
        self.raiz = self.inserir_no(
            self.raiz,
            valor
        )

    # Buscar um valor
    def buscar(self, valor):
        atual = self.raiz

        while atual is not None:

            if valor == atual.valor:
                return atual

            if valor < atual.valor:
                atual = atual.esquerda
            else:
                atual = atual.direita

        return None

    # Percurso em ordem
    def em_ordem(self, no):
        if no is not None:
            self.em_ordem(no.esquerda)

            print(
                no.valor,
                end=" "
            )

            self.em_ordem(no.direita)


# ==================================================
# PROGRAMA
# ==================================================

arvore = ArvoreAVL()

valores = [
    50, 28, 80, 18, 44, 62, 88,
    7, 23, 33, 45, 59, 65, 81, 89,
    13, 26, 41, 48, 61, 69, 86, 96
]

for valor in valores:
    arvore.inserir(valor)


print("Percurso em ordem:")
arvore.em_ordem(arvore.raiz)

print("\n")


# ==================================================
# MOSTRAR A RAIZ
# ==================================================

print("Raiz da árvore:", arvore.raiz.valor)

print(
    "Fator de balanceamento da raiz:",
    arvore.fator_balanceamento(arvore.raiz)
)
```
